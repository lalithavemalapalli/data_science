```python
df = demo.merge(behavior, on="ID", how="inner")
df = df.merge(credit, on="ID", how="inner")

print(df.shape)
print(df.head())
```

```python
df
```

##### Average credit card transactions

```python
df["Avg_CC_Spend"] = (
    df["cc_cons_apr"] + df["cc_cons_may"] + df["cc_cons_jun"]
) / 3
```

```python
df["Avg_CC_Spend"]
```

##### Average debit card transactions 

```python
df["Avg_DC_Spend"] = (
    df["dc_cons_apr"] + df["dc_cons_may"] + df["dc_cons_jun"]
) / 3
```

```python
df["Avg_DC_Spend"]
```

##### Total credit card transactions 

```python
df["Total_CC_Transactions"] = (
    df["cc_count_apr"] + df["cc_count_may"] + df["cc_count_jun"]
)
```

```python
df["Total_CC_Transactions"]
```

##### Total debit card transactions

```python
df["Total_DC_Transactions"] = (
    df["dc_count_apr"] + df["dc_count_may"] + df["dc_count_jun"]
)
```

```python
df["Total_DC_Transactions"]
```
