# Feature Engineering

## Project: Credit Card Consumption Prediction

### Overview

In this notebook, we will create new meaningful features from the existing customer transaction, spending, and investment columns.

Feature engineering helps improve analysis and machine learning performance by creating useful variables from raw data.

### What we will do in this notebook

- Create Total Credit Card Spend
- Create Total Debit Card Spend
- Create Total Transactions
- Create Total Investments
- Create Customer Spending Metrics
- Save the final feature-engineered dataset

### Why are we doing this?

- Raw columns may not directly show complete customer behavior.

- For example, monthly credit card spending columns show spending month-wise, but total credit card spend gives a better overall view of customer spending behavior.

- These new features will help the machine learning model make better predictions.

## Step 1: Import Required Libraries

We first import the required Python libraries.
- `pandas` is used for data manipulation.
- `numpy` is used for numerical operations.

These libraries help us create new features from the existing dataset.

```python
import pandas as pd
import numpy as np
```

## Step 2: Load the Cleaned Dataset

Now we load the cleaned dataset created in the data preprocessing notebook.
- Feature engineering should be performed on cleaned data.
- The cleaned dataset already has merged customer information, handled missing values, removed duplicates, and corrected data types.

```python
df = pd.read_csv("cleaned_credit_consumption_data.csv")
df
```

## Step 3: Check Available Columns

Before creating new features, we need to check the column names.
- Feature engineering depends on existing columns.
- By checking column names, we can confirm which spending, transaction, and investment columns are available in the dataset.

```python
df.columns.tolist()
```

## Step 4: Create Total Credit Card Spend

In this step, we create a new feature called `total_credit_card_spend`.
- The dataset contains monthly credit card spending columns.
- Instead of analyzing each month separately, we combine them to get the customer's total credit card spending.
- This feature helps us understand overall credit card usage.

```python
df["total_credit_card_spend"] = (
    df["cc_cons_apr"] +
    df["cc_cons_may"] +
    df["cc_cons_jun"]
)

df[["cc_cons_apr", "cc_cons_may", "cc_cons_jun", "total_credit_card_spend"]].head()
```

## Step 5: Create Total Debit Card Spend

In this step, we create a new feature called `total_debit_card_spend`.
- The dataset contains debit card spending for different months.
- By adding monthly debit card spending, we can understand the customer's total debit card usage.
- This helps compare debit card spending with credit card spending.

```python
df["total_debit_card_spend"] = (
    df["dc_cons_apr"] +
    df["dc_cons_may"] +
    df["dc_cons_jun"]
)

df[["dc_cons_apr", "dc_cons_may", "dc_cons_jun", "total_debit_card_spend"]].head()
```

## Step 6: Create Total Transactions

In this step, we create a new feature called `total_transactions`.
- Transaction count shows how frequently a customer uses cards.
- Customers with higher transaction counts may have higher credit card consumption.
- This feature combines both credit card and debit card transaction counts across available months.

```python
df["total_transactions"] = (
    df["cc_count_apr"] +
    df["cc_count_may"] +
    df["cc_count_jun"] +
    df["dc_count_apr"] +
    df["dc_count_may"] +
    df["dc_count_jun"]
)

df[[
    "cc_count_apr", "cc_count_may", "cc_count_jun",
    "dc_count_apr", "dc_count_may", "dc_count_jun",
    "total_transactions"
]].head()
```

## Step 7: Create Total Investments

In this step, we create a new feature called `total_investments`.
- Investment columns show the customer's financial activity.
- By combining all investment columns, we get the customer's overall investment value.
- This feature may help identify financially active customers.

```python
df["total_investments"] = (
    df["investment_1"] +
    df["investment_2"] +
    df["investment_3"] +
    df["investment_4"]
)

df[[
    "investment_1",
    "investment_2",
    "investment_3",
    "investment_4",
    "total_investments"
]].head()
```

## Step 8: Create Customer Spending Metrics

Now we create additional spending-based features.
- Spending metrics help us understand customer behavior more clearly.
- These features compare credit card usage, debit card usage, and overall spending activity.

```python
df["total_card_spend"] = (
    df["total_credit_card_spend"] +
    df["total_debit_card_spend"]
)

df["credit_spend_ratio"] = (
    df["total_credit_card_spend"] / df["total_card_spend"]
)

df["debit_spend_ratio"] = (
    df["total_debit_card_spend"] / df["total_card_spend"]
)

df[[
    "total_credit_card_spend",
    "total_debit_card_spend",
    "total_card_spend",
    "credit_spend_ratio",
    "debit_spend_ratio"
]].head()
```

## Step 9: Handle Infinite or Missing Values

While creating ratio features, some customers may have zero total card spending.

This can create infinite or missing values during division.
- Machine learning models cannot handle infinite values.
- So we replace infinite values with zero and fill any missing values created during feature engineering.

```python
df.replace([np.inf, -np.inf], 0, inplace=True)
df.fillna(0, inplace=True)
```

## Step 10: Check Newly Created Features

Now we verify whether all new features have been created successfully.

This step confirms that feature engineering was completed correctly and the new columns are available for analysis and machine learning.

```python
new_features = [
    "total_credit_card_spend",
    "total_debit_card_spend",
    "total_transactions",
    "total_investments",
    "total_card_spend",
    "credit_spend_ratio",
    "debit_spend_ratio"
]

df[new_features].head()
```

## Step 11: Save the Feature Engineered Dataset

After creating new features, we save the updated dataset.

Saving the feature-engineered dataset allows us to use it directly in the machine learning notebook without repeating these steps.

```python
df.to_csv("feature_engineered_credit_consumption_data.csv", index=False)
```
