## Step 6: Merge the Datasets

The information for each customer is spread across multiple files.

Machine learning models require a single consolidated dataset.

We will merge the datasets using the common customer identifier column (`ID`).

This will create one dataset containing:

- Customer demographics
- Customer behavior
- Credit card consumption

for every customer.

```python
merged_df = pd.merge(
    consumption_df,
    behavior_df,
    on="ID",
    how="inner"
)
```

```python
merged_df = pd.merge(
    merged_df,
    demographics_df,
    on="ID",
    how="inner"
)
```

## Step 9: Handle Missing Values

- After identifying missing values, we need to decide how to treat them.

- Machine learning algorithms cannot effectively handle large amounts of missing data.

- For numerical columns, we will replace missing values with the median.

- For categorical columns, we will replace missing values with the most frequent value (mode).

This helps preserve the dataset while improving data quality.

```python
num_cols = merged_df.select_dtypes(include=['int64', 'float64']).columns

for col in num_cols:
    merged_df[col] = merged_df[col].fillna(
        merged_df[col].median()
    )
```
