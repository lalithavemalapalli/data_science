## Step 4: Customer Demographics Analysis

In this step, we analyze customer demographic details such as age and gender.

- Demographic analysis helps us understand the customer base.

- It shows what type of customers are present in the dataset and whether age or gender may influence credit card consumption.

##### Age distribution Graph

```python
plt.figure(figsize=(8,5))
sns.histplot(df["age"], bins=30, kde=True)
plt.title("Age Distribution of Customers")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.show()
```

##### Gender distribution Graph

```python
plt.figure(figsize=(6,4))
sns.countplot(x="gender", data=df)
plt.title("Gender Distribution of Customers")
plt.xlabel("Gender")
plt.ylabel("Number of Customers")
plt.show()
```

## Step 5: Income Distribution Analysis

Now we analyze customer income.
- Income is an important factor in credit card spending.
- Customers with higher income may have higher purchasing power, so analyzing income helps us understand spending capacity.

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Income"], bins=30, kde=True)
plt.title("Income Distribution of Customers")
plt.xlabel("Income")
plt.ylabel("Number of Customers")
plt.show()
```

## Step 6: Credit Card Spending Pattern Analysis

In this step, we analyze the target variable `cc_cons`.
- `cc_cons` represents credit card consumption.
- Understanding its distribution helps us know whether most customers spend low, medium, or high amounts using credit cards.

```python
plt.figure(figsize=(8,5))
sns.histplot(df["cc_cons"], bins=30, kde=True)
plt.title("Credit Card Consumption Distribution")
plt.xlabel("Credit Card Consumption")
plt.ylabel("Number of Customers")
plt.show()
```

## Income vs Credit Card Consumption

Here we compare income with credit card consumption.

This helps us check whether customers with higher income also have higher credit card spending.

```python
plt.figure(figsize=(8,5))
sns.scatterplot(x="Income", y="cc_cons", data=df)
plt.title("Income vs Credit Card Consumption")
plt.xlabel("Income")
plt.ylabel("Credit Card Consumption")
plt.show()
```

## Step 7: Transaction Behavior Analysis

Now we analyze customer transaction behavior.
- Transaction count and spending behavior can strongly influence credit card consumption.
- Customers who perform more transactions may also have higher credit card usage.

```python
df["Total_CC_Transactions"] = df["cc_count_apr"] + df["cc_count_may"] + df["cc_count_jun"]

plt.figure(figsize=(8,5))
sns.histplot(df["Total_CC_Transactions"], bins=30, kde=True)
plt.title("Total Credit Card Transaction Distribution")
plt.xlabel("Total Credit Card Transactions")
plt.ylabel("Number of Customers")
plt.show()
```

## Step 8: Investment Trend Analysis

In this step, we analyze customer investment behavior.
- Investment activity may indicate financial stability.
- Customers who invest more may also have different credit card spending behavior compared to customers who invest less.

```python
investment_cols = ["investment_1", "investment_2", "investment_3", "investment_4"]

df["Total_Investment"] = df[investment_cols].sum(axis=1)

plt.figure(figsize=(8,5))
sns.histplot(df["Total_Investment"], bins=30, kde=True)
plt.title("Total Investment Distribution")
plt.xlabel("Total Investment")
plt.ylabel("Number of Customers")
plt.show()
```

## Step 9: Loan Analysis

Now we analyze whether loan-related features affect credit card consumption.
- Customers with loans or EMI payments may show different spending behavior.
- This analysis helps us understand the relationship between loan activity and credit card usage.

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="loan_enq", y="cc_cons", data=df)
plt.title("Credit Card Consumption by Loan Enquiry")
plt.xlabel("Loan Enquiry")
plt.ylabel("Credit Card Consumption")
plt.show()
```

## Step 10: Correlation Heatmap

A correlation heatmap shows the relationship between numerical variables.
- This helps us identify which features are strongly related to `cc_cons`.
- Highly correlated features can be useful for machine learning model building.

```python
numeric_df = df.select_dtypes(include=["int64", "float64"])

plt.figure(figsize=(15,10))
sns.heatmap(numeric_df.corr(), cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```
