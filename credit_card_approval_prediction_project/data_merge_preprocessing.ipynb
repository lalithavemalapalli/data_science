## Step 11: Handle Missing Values

From this dataset, the main missing values are usually found in `OCCUPATION_TYPE`.

This means the customer's occupation is not mentioned.

Instead of deleting these customers, we fill the missing occupation with `Unknown`.

Why?

Because occupation is a categorical column, and deleting many rows can remove useful customer records.

```python
application_df["OCCUPATION_TYPE"] = application_df["OCCUPATION_TYPE"].fillna("Unknown")

print("Missing values handled successfully")
```

## Step 15: Understand the STATUS Column

The `STATUS` column tells us the customer's repayment behavior.

### Meaning of STATUS values

| STATUS | Meaning | Customer Type |
|---|---|---|
| X | No loan for the month | Good |
| C | Loan closed | Good |
| 0 | Paid on time | Good |
| 1 | 30 days late | Bad/Risky |
| 2 | 60 days late | Bad/Risky |
| 3 | 90 days late | Bad/Risky |
| 4 | 120 days late | Bad/Risky |
| 5 | 150+ days late | Bad/Risky |

For our project, we will create a target variable:

- `0` means Good Customer
- `1` means Bad/Risky Customer

```python
credit_df["STATUS"].value_counts()
```

## Step 16: Create the Target Variable

Now we create the most important column for our machine learning project: `TARGET`.

Business logic:

- If the customer has status `1`, `2`, `3`, `4`, or `5`, then the customer has delayed payment. So we mark them as risky: `TARGET = 1`.
- If the customer has status `0`, `C`, or `X`, then the customer is treated as good: `TARGET = 0`.

This converts repayment history into a simple approval risk label.

```python
bad_status = ["1", "2", "3", "4", "5"]

credit_df["STATUS"] = credit_df["STATUS"].astype(str)
credit_df["TARGET"] = credit_df["STATUS"].apply(lambda x: 1 if x in bad_status else 0)

credit_df[["ID", "MONTHS_BALANCE", "STATUS", "TARGET"]].head()
```

## Step 17: Create One Credit Summary Per Customer

The credit file has many rows for the same customer because every row represents one month of payment history.

But the application file has one row for each customer profile.

So before merging, we summarize the credit history at customer level using `ID`.

For every customer, we create:

- `TARGET`: 1 if the customer was risky in any month, otherwise 0
- `TOTAL_MONTHS`: total months of credit history
- `MONTHS_ON_TIME`: months where payment was good
- `MONTHS_DELAYED`: months where payment was delayed
- `MAX_DELAY_STATUS`: highest delay level reached by the customer

```python
credit_summary = credit_df.groupby("ID").agg(
    TARGET=("TARGET", "max"),
    TOTAL_MONTHS=("MONTHS_BALANCE", "count"),
    MONTHS_ON_TIME=("STATUS", lambda x: x.isin(["0", "C", "X"]).sum()),
    MONTHS_DELAYED=("STATUS", lambda x: x.isin(bad_status).sum()),
    MAX_DELAY_STATUS=("STATUS", lambda x: max([int(i) for i in x if i in bad_status], default=0))
).reset_index()

credit_summary.head()
```

## Step 18: Check Target Distribution

Now we check how many customers are good and how many customers are risky.

This is important because if one class is much larger than the other, the machine learning model may become biased.

```python
credit_summary["TARGET"].value_counts()
```

## Step 19: Merge Application Data With Credit Summary

Now the main storytelling moment comes.

The bank wants both sides of the customer story:

1. Customer profile from `application_df`
2. Credit behavior from `credit_summary`

We merge both datasets using the common column `ID`.

We use `inner join` because we only want customers who are present in both files.

```python
merged_df = application_df.merge(credit_summary, on="ID", how="inner")

print("Merged Data Shape:", merged_df.shape)
merged_df.head()
```

## Step 20: Create Beginner-Friendly Columns

Some columns are difficult to understand directly.

For example:

- `DAYS_BIRTH` is stored as negative days
- `DAYS_EMPLOYED` is also stored as negative days for employed customers

So we create simple columns:

- `AGE`
- `EMPLOYMENT_YEARS`
- `INCOME_PER_FAMILY_MEMBER`
- `HAS_CHILDREN`

These columns will make EDA and machine learning easier.

```python
merged_df["AGE"] = (-merged_df["DAYS_BIRTH"] / 365).astype(int)

merged_df["EMPLOYMENT_YEARS"] = merged_df["DAYS_EMPLOYED"].apply(
    lambda x: 0 if x > 0 else int(abs(x) / 365)
)

merged_df["INCOME_PER_FAMILY_MEMBER"] = (
    merged_df["AMT_INCOME_TOTAL"] / merged_df["CNT_FAM_MEMBERS"]
).round(2)

merged_df["HAS_CHILDREN"] = merged_df["CNT_CHILDREN"].apply(lambda x: 1 if x > 0 else 0)

merged_df[["AGE", "EMPLOYMENT_YEARS", "INCOME_PER_FAMILY_MEMBER", "HAS_CHILDREN"]].head()
```
