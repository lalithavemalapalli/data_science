##### Here this line tells that , The columns Total charges looks like having all the numerics but sometimes pandas read it as string/text soo here 
- pd.to_numeric() funtion is used to convert all the values in the Total charges column to the numeric and,
- errors='coerce' this tells that is pandas finds the value which cannot be converted into the numbers then replace it with the NaN(missing value)

```python
df["Total Charges"] = pd.to_numeric(df["Total Charges"] , errors='coerce')
```

##### Here we are filling the missing values using function fillna() and the median of the Total charges column 

```python
df["Total Charges"] = df["Total Charges"].fillna(df["Total Charges"].median())
```

##### Here we are dividing the customers into 3 groups :
- 0-12 months - New customer (the customer who stayed for 12 months or less )
- 12-36 months - Medium customer (the customer who stayed more tham 12 months or less than 36 months)
- other - Long term customer (the customer who stayed more than 36 months)

```python
def customer_type(tenure):
    if tenure <= 12:
        return "New Customer"
    elif tenure <= 36:
        return "Medium Customer"
    else:
        return "Long Term Customer"

df["Customer_Type"] = df["Tenure Months"].apply(customer_type)
```

##### Here we are filtering only the churned customers because the main aim of our project is to find out the reason why the customers are churned from our company , so we are taking only the churned customers columns 

```python
churned_customers = df[df["Churn Label"] == "Yes"]
```

#### Filtering

### Grouping 

##### Now we need to know in total how many customers left the company and how many customers stayed in the company , we are grouping the customers into 2 groups left customers and stayed customers 

```python
churn_count = df["Churn Label"].value_counts()
churn_count
```

##### Here this result tells us that :
- Total customers churned - 1869
- Total chustomers stayed - 5174

##### Now lets us check what makes the customers churn whether the higher monthly charges made the customers churn lets see 

```python
avg_charges = df.groupby("Churn Label")["Monthly Charges"].mean()
avg_charges
```

##### so here this result tells us that the percentage of the churned customer based on the Monthly charges 
- churned cutomers : 74.441332
- stayed customers / Not churned customers : 61.265124

##### Now let us group the churned and stayed cutomers by the Contract 

```python
contract_churn = df.groupby(["Contract", "Churn Label"]).size()

print(contract_churn)
```

#### Pivot Tables 

##### 1. Churn by Contract 

```python
pivot_contract = pd.pivot_table(
    df,
    index="Contract",
    columns="Churn Label",
    values="CustomerID",
    aggfunc="count"
)

print(pivot_contract)
```

##### 2. Churn by Payment Method

```python
pivot_payment = pd.pivot_table(
    df,
    index="Payment Method",
    columns="Churn Label",
    values="CustomerID",
    aggfunc="count"
)

print(pivot_payment)
```

##### 3. Average Monthly charges by Contract

```python
pivot_avg_charges = pd.pivot_table(
    df,
    index="Contract",
    values="Monthly Charges",
    aggfunc="mean"
)

print(pivot_avg_charges)
```
