## Project Title : Customer Churn Analysis and Prediction System

### BUSINESS PROBLEM
Imagine a telecom company is losing customers every month.
The company does not know:
- why customers are leaving
- which customers are likely to leave
- which services cause dissatisfaction
- how to reduce customer loss
##### Customer loss is called: “Customer Churn”
Every customer who leaves the company causes:
- revenue loss
- profit reduction
- business instability
So the company wants a data analyst to analyze customer behavior and identify the major reasons behind churn.
That is where this project starts.

## FLOW OF THE PROJECT

#### STEP 1 → Understanding the Business Problem
The telecom company has thousands of customers using:
- phone services
- internet services
- streaming services
- online payment systems

But many customers are leaving the company unexpectedly.
The company wants answers for questions like:
- Which customers are leaving more?
- Why are they leaving?
- Which contract type has highest churn?
- Do high monthly charges increase churn?
- Are long-term customers more loyal?
- Which payment method causes more churn?
To solve this problem, customer data was collected.

#### STEP 2 → Data Collection
The dataset contains customer information such as:
- gender
- senior citizen status
- tenure
- internet service
- monthly charges
- payment methods
- contract type
- total charges
- churn status

The main target column is: Churn

Where:
- Yes → customer left company
- No → customer stayed

#### STEP 3 → Data Cleaning and Preparation
Before analysis, the dataset was cleaned using Excel and Python.

The cleaning process included:
- handling missing values
- checking duplicate records
- correcting data types
- understanding categorical columns
- preparing data for analysis

This step ensured the dataset was accurate and reliable.

#### STEP 4 → Exploratory Data Analysis (EDA) 
This is the heart of the project.

Using Python libraries like:
- pandas
- matplotlib
- seaborn

different visualizations were created to identify customer churn patterns.

#### KEY ANALYSIS PERFORMED

1. Churn Distribution Analysis
- First, the number of customers who stayed and left was analyzed.
- This helped understand the overall churn rate of the company.

Insight: A significant number of customers were leaving the company, indicating a serious business issue.

2. Contract Type vs Churn

Customers with:
- month-to-month contracts
- yearly contracts
- two-year contracts were compared.

Insight:
- Customers with month-to-month contracts had the highest churn rate because they had less long-term commitment.
- Long-term contract customers were more loyal.

3. Monthly Charges vs Churn
- Customer monthly bills were analyzed.

Insight:
- Customers paying higher monthly charges were more likely to leave the company.
- This suggests pricing dissatisfaction.

4. Tenure vs Churn
- Customer duration with the company was studied.

Insight:
- New customers churned more frequently, while long-term customers showed higher loyalty.
- This indicates the importance of improving early customer experience.

5. Payment Method Analysis

Different payment methods were analyzed:
- electronic check
- mailed check
- bank transfer
- credit card

Insight:
- Customers using electronic check showed higher churn behavior compared to other payment methods.

## Business Insights 

After analyzing the data , several important business insights were found 
- Reason 1 : High monthly charges increase churn probability.
- Reason 2 : Month-to-month contract customers are at highest risk.
- Reason 3 : New customers are more likely to leave.
- Reason 4 : Electronic check users show unstable customer behavior.
- Reason 5: Fiber optic customers have higher dissatisfaction levels.

## Solutions for the above insights 

Based on the analysis, the following solutions were suggested to the telecom company:
- Provide discounts for long-term contracts to reduce churn.
- Improve customer onboarding experience for new customers.
- Offer customized plans for high monthly charge customers.
- Improve customer support for fiber optic users.
- Introduce loyalty rewards for long-term customers.
