#### Here we are visualizing the churned and stayed customers besacuse if the churned customers are high the company is running in lose 

```python
sns.countplot(x='Churn Label', data=df)

plt.title("Customer Churn Count")

plt.show()
```

#### Here we are checking in which contract type the churned customers are high 
- More customers are churned those who are having the month-to-month contract 

```python
sns.countplot(x='Contract', hue='Churn Label', data=df)

plt.title("Contract Type vs Churn")

plt.show()
```

#### Here we are checking the churned customers based on the payment methods 
- More customers are churned due to the electronic check payment method 

```python
plt.figure(figsize=(10,5))

sns.countplot(x='Payment Method', hue='Churn Label', data=df)

plt.xticks(rotation=45)

plt.show()
```

#### Here we are understanding the charge distribution among the customers 

```python
sns.histplot(df['Monthly Charges'], bins=30)

plt.title("Monthly Charges Distribution")

plt.show()
```

#### Here we are understanding whether the high charges paying customers are leaving 

```python
sns.boxplot(x='Churn Label', y='Monthly Charges', data=df)

plt.show()
```

#### Here we are checking whether the new customers are leaving or long term customers are leaving 

```python
sns.boxplot(x='Churn Label', y='Tenure Months', data=df)

plt.show()
```
